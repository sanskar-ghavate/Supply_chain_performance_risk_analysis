```python
# creating copy of original dataset
df_clean = df.copy()

df_clean.columns
```

```python
columns_to_drop = ['Product Description', 'Order Zipcode','Customer Fname','Customer Lname','Customer Email','Customer Password',
                   'Customer Street','Customer Zipcode','Product Image']

df_clean = df_clean.drop(columns=columns_to_drop)

print(df_clean.shape)
```

```python
# check missing values after removing unnecessary columns.
df_clean.isnull().sum().sum()
```

```python
#renaming the columns

new_columns = []

for col in df_clean.columns:
    col = col.strip()
    col = col.lower()
    col = col.replace(" ","_")
    col = col.replace("(","")
    col = col.replace("(","")
    new_columns.append(col)

df_clean.columns = new_columns

df_clean.columns.tolist()
```

```python
df_clean.columns.duplicated().sum()
```

```python
df_clean.rename(columns={'order_date_dateorders)':'order_date_dateorders',
                         'shipping_date_dateorders)': 'shipping_date_dateorders',
                         'days_for_shipping_real)':'days_for_shipping_real',
                         'days_for_shipment_scheduled)':'days_for_shipment_scheduled'},inplace=True)
```

```python
#Convert date columns to datetime

df_clean['order_date_dateorders'] = pd.to_datetime(df_clean['order_date_dateorders'],format='%m/%d/%Y %H:%M')

df_clean['shipping_date_dateorders'] = pd.to_datetime(df_clean['shipping_date_dateorders'],format='%m/%d/%Y %H:%M')
```

```python
# check and negative and zero values

columns_to_check = ['sales','order_item_quantity','order_item_product_price',
                    'order_profit_per_order','days_for_shipping_real','days_for_shipment_scheduled']

for col in columns_to_check:
    print(f"{col}")
    print("Negative Values:",(df_clean[col]<0).sum())
    print("Zero Values:",(df_clean[col]==0).sum())
```

```python
# checking shipping dates earlier than order dates

invalid_shipping_dates = df_clean[df_clean['shipping_date_dateorders'] < df_clean['order_date_dateorders']]

print(len(invalid_shipping_dates))
```

```python
# Check duplicate order item IDs

print(df_clean['order_item_id'].duplicated().sum())
```

```python
df_clean['shipping_delay_days'] = (df_clean['days_for_shipping_real'] - df_clean['days_for_shipment_scheduled'])

pd.crosstab(df_clean['delivery_status'], df_clean['shipping_delay_days'])
```

```python
# checking orders having multiple delivery status

status_check = df_clean.groupby('order_id')['delivery_status'].nunique()

print((status_check > 1).sum())
```

```python
# checking orders containing multiple shipping modes

shipping_check = df_clean.groupby('order_id')['shipping_mode'].nunique()

print((shipping_check > 1).sum())
```

```python
#checking missing values and dataset dimensions

print("Total Rows:", df_clean.shape[0])
print("Total Rows:", df_clean.shape[1])
print("Missing Values:", df_clean.isnull().sum().sum())
print("Duplicate Order Items:",df_clean['order_item_id'].duplicated().sum())
```

```python
# Export cleaned dataset

df_clean.to_csv(r"C:\Users\sansk\OneDrive\Desktop\Study\Projects\Final Projects\Supply_Chain_Performance_Analytics\Data\cleaned\supply_chain_cleaned.csv",
                index = False)

print("Cleaned dataset exported successfully")                 
```
